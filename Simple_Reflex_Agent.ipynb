```python
class SimpleReflexAgent:
    def __init__(self, rules):
        self.rules = rules

    def interpret_input(self, percept):
        temperature = percept["temperature"]
        if temperature < 20:
            state = "COLD"
        else:
            state = "NOT_COLD"
        return state

    def rule_match(self, state):
        for rule in self.rules:
            if rule["condition"] == state:
                return rule["action"]
        return "NO_ACTION"

    def act(self, percept):
        state = self.interpret_input(percept)
        action = self.rule_match(state)
        return state, action


rules = [
    {"condition": "COLD", "action": "HEATER_ON"},
    {"condition": "NOT_COLD", "action": "HEATER_OFF"}
]

agent = SimpleReflexAgent(rules)

test_temperatures = [16, 19, 20, 24, 30]

for temp in test_temperatures:
    percept = {"temperature": temp}
    state, action = agent.act(percept)

    print("--------------------------")
    print("Temperature:", temp, "°C")
    print("Interpreted State:", state)
    print("Selected Action:", action)
```
