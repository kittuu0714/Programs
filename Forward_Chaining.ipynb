```python
rules = [
    ("R1", ["CAR_NOT_STARTING", "LIGHTS_DIM"], "BATTERY_WEAK"),
    ("R2", ["BATTERY_WEAK", "ENGINE_NOT_CRANKING"], "BATTERY_PROBLEM"),
    ("R3", ["BATTERY_PROBLEM"], "CHECK_BATTERY"),
    ("R4", ["FUEL_GAUGE_EMPTY"], "NO_FUEL"),
    ("R5", ["CAR_NOT_STARTING", "NO_FUEL"], "REFUEL_CAR")
]

facts = {
    "CAR_NOT_STARTING",
    "LIGHTS_DIM",
    "ENGINE_NOT_CRANKING"
}

initial_facts = facts.copy()
applied_rules = []
inferred_facts = []

changed = True

while changed:
    changed = False

    for name, conditions, conclusion in rules:
        if all(c in facts for c in conditions):
            if conclusion not in facts:
                facts.add(conclusion)
                inferred_facts.append(conclusion)
                applied_rules.append(name)
                changed = True

print("\nFORWARD CHAINING RESULT")
print("Initial Facts:", initial_facts)
print("Inferred Facts:", inferred_facts)
print("Applied Rules:", applied_rules)
```
