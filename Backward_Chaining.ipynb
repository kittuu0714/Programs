```python
rules = [
    ("R1", ["CAR_NOT_STARTING", "LIGHTS_DIM"], "BATTERY_WEAK"),
    ("R2", ["BATTERY_WEAK", "ENGINE_NOT_CRANKING"], "BATTERY_PROBLEM"),
    ("R3", ["BATTERY_PROBLEM"], "CHECK_BATTERY"),
    ("R4", ["FUEL_GAUGE_EMPTY"], "NO_FUEL"),
    ("R5", ["CAR_NOT_STARTING", "NO_FUEL"], "REFUEL_CAR")
]

facts = {
    "CAR_NOT_STARTING",
    "LIGHTS_DIM",
    "ENGINE_NOT_CRANKING"
}

def backward_chaining(goal, visited=None):
    if goal in facts:
        return True

    if visited is None:
        visited = set()

    if goal in visited:
        return False

    visited.add(goal)

    for name, conditions, conclusion in rules:
        if conclusion == goal:
            if all(backward_chaining(c, visited.copy())
                   for c in conditions):
                return True

    return False


goal = "CHECK_BATTERY"

print("\nBACKWARD CHAINING RESULT")
print("Initial Facts:", facts)
print("Goal:", goal)

if backward_chaining(goal):
    print("Goal can be proved from given facts.")
else:
    print("Goal cannot be proved from given facts.")
```
